## Using the CLI

Change to the home directory and check which file directory you are working in

````bash
cd ~
pwd
````

Check what is there (probably nothing)

````bash
ls
````

Set the time and date to your local time

````bash
export TZ=US/Central
````

To use the AWS command to list your s3 buckets:

````bash
aws s3 ls
````

## Using S3 with the CLI

Use a variable for the bucket name and display it. Be sure to replace `learner` with your user username (for example `user1766969`)

````bash
BUCKET=essentials-test-learner-$(date +%F)
echo $BUCKET
````

Let's list the contents of your s3 bucket:

````bash
aws s3 ls s3://$BUCKET
````

Looks like there is nothing there right now. So let us create a text file called `hemingway.txt` and populate it with a nonsensical quote. 

````bash
echo "The world breaks everyone" > hemingway.txt
````

Check the file just to be sure

````bash
ls
````

````bash
cat hemingway.txt
````

Now let's move the file into our s3 bucket. 

````bash
aws s3 mv ./hemingway.txt s3://$BUCKET
````

Check the it was a move (not a copy) by checking the file is gone

````bash
ls
````

````bash
aws s3 cp s3://$BUCKET/hemingway.txt .
````

Check that we got the file back

````bash
ls
````

````bash
cat hemingway.txt
````

Cleanup after ourselves

````bash
rm hemingway.txt
````

```{admonition} Danger
:class: Danger

The `aws s3 mv` command permanently deletes files. They cannot be recovered!
```
