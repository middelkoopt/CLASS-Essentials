Open the cloud shell (See: intro_to_cli)

Delpoy blob storage in Azure

Define some environment variables

```python
RG="demo_rg"
LOCATION=eastus
BLOB_CON="demo_blob_cont"
STGE_ACCT="demo_stor_ac"
CTR="demo_stge_ctr"
```

Create resource group

az group create --name $RG --location $LOCATION

Create storage account

All storage resources need to belong to a storage account.

```python
az storage account create --name $STGE_ACCT --resource-group $RG --location $LOCATION --sku Standard_ZRS --encryption-services blob
```

Create storage container

The following example uses your Azure AD account to authorize the operation to create the container. Before you create the container, assign the Storage Blob Data Contributor role to yourself. Even if you are the account owner, you need explicit permissions to perform data operations against the storage account.

az ad signed-in-user show --query objectId -o tsv | az role assignment create \
    --role "Storage Blob Data Contributor" \
    --assignee @- \
    --scope "/subscriptions/<subscription>/resourceGroups/$RG/providers/Microsoft.Storage/storageAccounts/$STGE_ACCT"

Now create the storage Container 

az storage container create --account-name $STGE_ACCT --name $CTR --auth-mode login

Now we will upload a file from your local machine

Create a file to upload

```python
echo "Demo text" > demo_file
```

Copy demo_file to your blob.

az storage blob upload --account-name $STGE_ACCT --container-name $CTR --name demo_file --file demo_file --auth-mode login

Verify that the blob is there by Listing the blobs in the container. 

```python
az storage blob list --account-name $STGE_ACCT --container-name $CTR --output table --auth-mode login
```

Download the file we just uploaded.

az storage blob download --account-name $STGE_ACCT --container-name $CTR --name demo_file --file demo_file_download file --auth-mode login

In Cloud Shell type the following command:

```python
ls demo_file_download
```

Verify that the downloaded file is the same file we uploaded.

```python
diff demo_file demo_file_download
```

Clean up

We will clean up the resources in this lesson in the reverse order that we created them.  

## Blob
Delete a blob

```python
az storage blob delete --account-name $STGE_ACCT --container-name $CTR --name demo_file --auth-mode login
```

Verify the blob was deleted

```python
az storage blob list --account-name $STGE_ACCT --container-name $CTR --output table --auth-mode login
```

## Storage Account
Delete the Storage Account.

```python
az storage account delete -n $STGE_ACCT -g RG
```

Verify that the Storage Account has been deleted.

```python
az storage account list | grep $STGE_ACCT
```

## Source Group
By deleting the Resource Group, you will delete all resouce in that group.

The --no-wait parameter keeps the CLI from blocking while the deletion takes place.

```python
az group delete --name $RG --no-wait
```
