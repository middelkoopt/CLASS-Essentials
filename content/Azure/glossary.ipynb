# Glossary

* **Active Directory** - This is were your identity is defined.  This is typically setup by the institution that you belong to.
* **Subscription** - Resources and services in Azure are created within a Subsrciption and this acts as your billing account. 
* **Account** - The Azure account (email) associated with an individual.  This is the identity that you use to access console.
* **Resource Group** - A resource group is a *container* (in the literal sense not to be confused with a Docker container for example) that holds related resources that you want to manage as a group.  Resources are deployed in a Resource Group.
