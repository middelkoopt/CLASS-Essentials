## Enumerating Existing Resources

We will first use the method from the previous lesson (see if you can do this on your own first) to check which resources are currently running so we can verify that they get removed.

````bash
gcloud asset search-all-resources --asset-types='compute.googleapis.com/Disk,compute.googleapis.com/Instance,storage.googleapis.com/Bucket' --format='value(name)'
````

````bash
gcloud asset search-all-resources --format='value(name)' |grep -vE '/(services|networks|routes|subnetworks|firewalls|serviceAccounts)/' | sort
````

## Cleanup Buckets

First set the environment variables and verify them.  If they are previously set you can just verify them.

````bash
ACCOUNT=$(gcloud config list --format='value(core.account)')
PROJECT=$(gcloud config list --format='value(core.project)')
BUCKET="essentials-${USER}-*"
REGION="us-west2"
````

````bash
echo "Account: $ACCOUNT, Project: $PROJECT, Region: $REGION"
echo "Bucket: $BUCKET"
````

## Cleanup Buckets

We will now list the buckets that we will be deleting. Note the wild-card (GLOB) from above will match all dates (and all strings).

````bash
gsutil ls -b "gs://$BUCKET"
````

**WARNING: All of the above buckets will be deleted if the following commands are run**

**Delete** all the objects first

````bash
gsutil -m rm -r "gs://$BUCKET/*"
````

Delete the empty buckets

````bash
gsutil rb "gs://$BUCKET/"
````

## Enumerating Resources
