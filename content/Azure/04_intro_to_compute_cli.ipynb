**Objectives:**
 * Allocate a virtual machine in Azure.

Open the cloud shell (See: intro_to_cli)

Define some environment variables

```python
RG="demo_rg"
LOCATION=eastus
VM="demo_vm"
STGE_ACCT="demo_stor_ac"
CTR="demo_stge_ctr"
```

Create resource group

```python
az group create --name $RG --location $LOCATION
```

Create a VM

If this is the fist time running this command (or your ~/.ssh dir does not have an id_rsa key pair).  It will generate a new key pair id_rsa in ~/.ssh. 

```python
az vm create --resource-group $RG \
  --name $VM \
  --image UbuntuLTS \
  --admin-username demouser \
  --generate-ssh-keys \
  --output json \
  --verbose 
```

To create a VM with an existing pair first create a new key pair

```python
ssh-keygen
```

You will be prompted to "Enter file in which to save the key", type "id_rsa".  This will create a key pair in the current dir.

Here you should enter a passphrase (a series of words) to protect the key. You will be asked to re-enter the passphrase as shown below:

```python
Enter same passphrase again:
```

After successfully retyping the passphrase the key generation process will continue showing similar message as follows (the fingerprint and art will look different):

Note: The following using "--ssh-key-values" to define which keys to use.

```python
az vm create --name $VM 
  --resource-group $RG 
  --image UbuntuLTS 
  --admin-username demouser 
  --ssh-key-values id_rsa.pub
```

Show VM details (all)

```python
az vm show --name $VM --resource-group $RG
```

Login into VM

Get the public IP address from the output above.
If you generated a new key pair using "--generate-ssh-keys", login with the following command.

```python
ssh <PUBLIC_IP_ADDRESS>
```

If you used the command with "--ssh-key-values id_rsa.pub" to use existing keys, use the following command to logiin:

```python
ssh -i <path to ssh key>/id_rsa <PUBLIC_IP_ADDRESS>
```
