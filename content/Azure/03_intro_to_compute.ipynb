## Azure Virtual Machines

## Create a VM

From the hamburger menu in the upper left of the top blue bar, click *Virtual Machines*. On the *Virtual machines* 
page click Create (+) on the menu in upper left of the page. Choose *Virtual machines*.
In the middle of the page under the "Create Virtual Machine" heading there are menu options from left to right:  
Basic, Disk, Networking, Management, Advanced, Tags, Review & Create.  We will start with the Basic tab (default).

**Basic**

*  Check subscription
  *  Use the pull down menu to change subscriptions if needed
*  Resource Group
  *  Click the down arrow in the Resource Group window and choose 
    the resource group "essentials" that we created earlier.
*  Virtual machine name
  *  Enter "essentials-VM"
*  Region
  * Make sure you are in (US) East US
*  Image
  *  Use default image (Ubuntu 20.04 - Gen2)
*  Size
  *  Standard_D2s_v3 (default)

![vm-create](images/01_CLASS_Azure_VM_basic_01.png)

*  Authentication type 
  *  SSH public key (default)
*  Username
  *  Choose a username (default: azeruser)
*  SSH public key source
  *  Select Use existing key stored in Azure
*  Stored keys
  *  Click on "Select a key" and choose "essentials-public-key" that 
    we created earlier in the setup episode with the "SSH key" service.

Leave the rest of the basic tab with the default settings

![vm-create](images/01_CLASS_Azure_VM_basic_02.png)

Now we can provision the VM.  The Management, Monitoring, 
Advanced, and Tags menu options across the top are outside the 
scope of this tutorial so we are now ready to click Review & Create.

## Connect to the VM Instance

To connect to the *VM Instance* we will need to get the username and
IP address for the machine we just created. First, navigate to the 
VM overview page if not there already.

To connect to the machine we will use the cloud shell in the 
web console.  But first we need to get the username and
IP address for the machine we just created. On the VM overview page, 
locate the "Public IP address" and hover your mouse over the number. 
This will bring up an icon to the right where you can copy this number 
to your clipboard. (image)

Open up a cloud shell by clicking on the **cloud Shell** icon 
(button the looks like a box with a greater than sign followed 
by an underscore) on the top blue bar if the cloud shell is not 
already open.

In the cloud shell type "ssh", then a space, then the username 
(azureuser by default), followed by an "@" sign. Now paste 
(OS dependent) the public IP number directly after the "@" sign and 
hit return.

````python
ssh <username>@<Public IP address>
````

If this is your first time logging in, you will be prompted with The 
authenticity of host 'IP' can't be established. Type 'yes' and this 
will add the host to the ~/.ssh/*know_hosts file. After doing this 
once you will not be prompted again (unless the host key changes or 
the host is removed from the ~/.ssh/know_hosts file.

````python
ssh azureuser@10.10.10.10
The authenticity of host '10.10.10.0 (10.10.10.0)' can't be established.
ECDSA key fingerprint is SHA256:F/3poSNTVVNbfSp3Kc7zBGwiGzt3qwrmjiyfI1ZHnX4.
Are you sure you want to continue connecting (yes/no)?
````

Now you should be logged into your VM.  You will notice that your
command prompt has changed to "azureuser@essentials-VM".
